## The data

```python
# ---- Setup: run this cell first ----
!pip install -q duckdb

import glob
import os

import duckdb
import pandas as pd

DATA_URL = "https://github.com/MJMortensonWarwick/adminaeaiv/releases/download/dbsession/bike-hire-practice.zip"

DATA_DIR = "/content/bikes"
# Download only if no CSV files are there yet. The CSVs may sit inside a folder
# within the zip; the search below looks in sub-folders too.
if DATA_URL and not glob.glob(os.path.join(DATA_DIR, "**", "*.csv"), recursive=True):
    import urllib.request
    import zipfile
    os.makedirs(DATA_DIR, exist_ok=True)
    urllib.request.urlretrieve(DATA_URL, "/content/bikes.zip")
    zipfile.ZipFile("/content/bikes.zip").extractall(DATA_DIR)

FILES = {"trips": "trips.csv", "bikes": "bikes.csv", "stations": "stations.csv"}

tables, missing = {}, []
for name, filename in FILES.items():
    found = glob.glob(os.path.join(DATA_DIR, "**", filename), recursive=True)
    if found:
        tables[name] = pd.read_csv(found[0])
    else:
        missing.append(filename)

if missing:
    found_csvs = sorted(os.path.basename(f) for f in glob.glob(os.path.join(DATA_DIR, "**", "*.csv"), recursive=True))
    raise FileNotFoundError(
        f"Missing files: {missing}. CSV files found under {DATA_DIR}: {found_csvs}. Ask for help."
    )

# Make every table available as a DuckDB table, so you can use SQL or pandas.
con = duckdb.connect()
for name, df in tables.items():
    con.register(name, df)


def q(sql):
    """Run SQL against the tables above and return a pandas DataFrame."""
    return con.sql(sql).df()


for name, df in tables.items():
    print(f"{name}: {df.shape[0]:,} rows, {df.shape[1]} columns")
    print("   ", list(df.columns))
```

## 2. Clean and prepare

*Decide what to do about each problem you found, say why, and do it. Short cells that you can re-run are better than one long cell.*

```python
# Clean tables in-place using direct SQL statements
con.execute("""
    CREATE OR REPLACE TABLE bikes AS
    SELECT
        bike_id,
        CASE
            WHEN LOWER(TRIM(bike_type)) LIKE '%standard%' THEN 'Standard'
            WHEN LOWER(TRIM(bike_type)) LIKE '%electric%' OR LOWER(TRIM(bike_type)) LIKE '%e-bike%' THEN 'Electric'
            ELSE bike_type
        END AS bike_type,
        purchase_year
    FROM bikes
""")

con.execute("""
    CREATE OR REPLACE TABLE trips AS
    SELECT
        trip_id,
        bike_id,
        start_station_id,
        end_station_id,
        start_time,
        duration_min,
        CASE
            WHEN LOWER(TRIM(user_type)) = 'casual' THEN 'Casual'
            WHEN LOWER(TRIM(user_type)) = 'member' THEN 'Member'
            ELSE user_type
        END AS user_type
    FROM trips
""")

# Quick confirmation checks
print("Cleaned Bike Types:")
display(con.sql("SELECT DISTINCT bike_type FROM bikes").df())

print("\nCleaned User Types:")
display(con.sql("SELECT DISTINCT user_type FROM trips").df())
display(q("SELECT DISTINCT bike_type FROM bikes"))
display(q("SELECT DISTINCT zone FROM stations"))
display(q("SELECT DISTINCT user_type FROM trips"))
```

```python
# 1. Clean the original DataFrames inside the 'tables' dictionary directly
import pandas as pd

# Standardize bike_type
bikes_df = tables["bikes"]
bikes_df["bike_type"] = bikes_df["bike_type"].astype(str).str.strip().str.lower().map({
    'standard': 'Standard',
    'e-bike': 'Electric',
    'electric': 'Electric'
}).fillna(bikes_df["bike_type"])

# Standardize user_type
trips_df = tables["trips"]
trips_df["user_type"] = trips_df["user_type"].astype(str).str.strip().str.lower().map({
    'casual': 'Casual',
    'member': 'Member'
}).fillna(trips_df["user_type"])

# 2. Re-register the cleaned DataFrames to DuckDB
con.register("bikes", bikes_df)
con.register("trips", trips_df)

# 3. Verify the final distinct types
print("Distinct Bike Types:")
display(con.sql("SELECT DISTINCT bike_type FROM bikes").df())

print("\nDistinct User Types:")
display(con.sql("SELECT DISTINCT user_type FROM trips").df())
```

## 3. Join and aggregate

*Join the tables and calculate the answer. Note how many rows you expect after each join, then check.*

```python
### 1. How to join the tables and calculate the answer?
# To answer the manager's brief, we need columns from three separate tables:
#   - `bike_type` from the `bikes` table
#   - `zone` (specifically for the start station) from the `stations` table
#   - `duration_min` and `trip_id` from the `trips` table
#
# We will write an SQL query that joins `trips` to `bikes` using `bike_id`, and
# joins `trips` to `stations` using `start_station_id` = `station_id`.
# Finally, we will use `GROUP BY bike_type, zone` to calculate both the average duration
# and the total trip count.

### 2. Which JOIN is best to use and why?
# An INNER JOIN is the best choice here because:
#   - We only want to analyze trips where we have complete information for both the bike's type and the station's zone.
#   - If there were any trips with missing/invalid `bike_id` or `start_station_id` values (or IDs not present in our dimension tables),
#     an INNER JOIN would naturally exclude them, preventing null or corrupted entries from skewing our averages.
#   - Since every trip in this dataset has valid matching foreign keys, an INNER JOIN is clean, performant, and safe.

# Let's execute the join and inspect the results:
query = """
SELECT
    b.bike_type,
    s.zone AS start_zone,
    COUNT(t.trip_id) AS total_trips,
    ROUND(AVG(t.duration_min), 2) AS avg_duration_min
FROM trips AS t
INNER JOIN bikes AS b
    ON t.bike_id = b.bike_id
INNER JOIN stations AS s
    ON t.start_station_id = s.station_id
GROUP BY b.bike_type, s.zone
ORDER BY b.bike_type, s.zone;
"""

results_df = q(query)
display(results_df)
```
